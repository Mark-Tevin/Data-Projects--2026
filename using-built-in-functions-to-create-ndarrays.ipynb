#### 1. Create a Numpy array of zeros with a desired shape

```python
# Create a 3 x 4 ndarray full of zeros.
X = np.zeros((3,4), dtype=int)

# we print x
print()
print('X = \n', X)
print()

# we print information about x
print('X has dimension:', X.shape)
print('X is an object of type:', type(X))
print('The elements in X are of type:', X.dtype)
```

#### 6.Create a Numpy array of evenly spaced values in a given range, using arange(stop_val)
[https://numpy.org/doc/stable/reference/generated/numpy.arange.html](http://)

```python
# we create a rank 1 ndarray that has sequential integers from 0 to 9
x = np.arange(10)

# we print the ndarray
print()
print('x = ', x)
print()

# we print information about the ndarray
print('x has dimensions:', x.shape)
print('x is an object of type:', type(x))
print('The elements in x are of type:', x.dtype)
```

#### 8.Create a numpy array using linspace(start, stop, n), with stop inclusive
###### References
[Reference/Docs](https://numpy.org/doc/stable/reference/generated/numpy.linspace.html)

```python
# we create a rank 1 ndarray that has 10 integers evenly spaced between 0 and 25
x = np.linspace(0,25,10)

# print the ndarray
print()
print('x = \n', x)
print()

# we print information about ndarray
print('x has dimensions:', x.shape)
print('x is an object of type:', type(x))
print('The elements in x are of type:', x.dtype)
```

#### Create a Numpy array using linspace(start, stop, n), with stop excluded

```python
# Creating a rank 1 ndarray that has 10 integers evenly spaced between 0 and 25,
# with 25 excluded
x = np.linspace(0,25,10, endpoint = False)

# Print the ndarray
print()
print('x = ', x)
print()

# we print information about the ndarray
print('x has dimensions:', x.shape)
print('x is an object of type:', type(x))
print('The element in x are of type:', x.dtype)
```

#### 9.Create a Numpy array by feeding the output of arange() function to the reshape() function.
Here arange() function will give you a 1-D array, whereas the reshape() function will convert that 1-D array into a desired shape.
NB: The size() of the final output must be as same as size of initial 1-D array.

```python
# We create a rank 1 ndarray with sequential integers from 0 to 19
x = np.arange(20)

# we print x
print()
print('Original x = ', x)
print()

# we reshape x into a 4 x 5 ndarray
x = np.reshape(x, (4,5))

# we print the reshaped x
print()
print('Reshaped x = \n', x)
print()

# we print the information about reshaped x
print('x has dimensions:', x.shape)
print('x is an object of type:', type(x))
print('The elements in x are of type:', x.dtype)
```

#### 10 Create a Numpy array by calling the reshape() function from the output of arange() function.

###### changing in arguments of reshape()


```python
# We create a rank 1 ndarray with sequential integers from 0 to 19 and
# reshape it to a 4 x 5 array
Y = np.arange(20).reshape(4, 5)

# we print Y
print()
print('Y = \n', Y)
print()

# we peint information about Y
print('Y has dimensions:', Y.shape)
print('Y is an object of type:', type(Y))
print('The elements in Y are of type:', Y.dtype)
```

#### 11. Create a rank 2 Numpy array by using the reshape() function

```python
# We create a rank 1 ndarray with 10 integers evenly spaced between 0 and 50,
# with 50 excluded. we then reshape it to a 5 x 2 ndarray
X = np.linspace(0,50,10, endpoint=False).reshape(5,2)

# we print x
print()
print('X = \n', X)
print()

# we print information about x
print('X has dimensions:', X.shape)
print('X is an object of type:', type(X))
print('The elements in X are of type:', X.dtype)
```

#### 12. Create a Numpy array using numpy.random.random() function

```python
# We create a 3 x 3 ndarray with random floats in the half-open interval [0.0, 1.0].
X = np.random.random((3,3))

# we print x
print()
print('X = \n', X)
print()

# we print information about x
print('X has dimensions:', X.shape)
print('X in an object of type:', type(X))
print('The elements in X are of type:', X.dtype)
```

#### 13.Create a Numpy array using the numpy.random.randint() function

```python
# We create a 3 x 2 ndarray with random integers in the half-open interval [4, 15].
X = np.random.randint(4,15,size=(3,2))

# we print x
print()
print('X = \n', X)
print()

# we print information about X
print('X has dimensions:', X.shape)
print('X is an object of type:', type(X))
print('The elements in X are of type:', X.dtype)
```

#### 14.Create a Numpy array of "Normal" distributed random numbers, using the numpy.random.normal() function.

```python
# We create a 1000 x 1000 ndarray of random floats drawn from normal (Gaussian) distribution
# with a mean of zero and a standard deviation of 0.1.
X = np.random.normal(0, 0.1, size=(1000,1000))

# we print X
print()
print('X = \n', X)
print()

# we print information about x
print('X has dimensions:', X.shape)
print('X is an object of type:', type(X))
print('The elements in X are of type:', X.dtype)
print('The elements in X havea mean of:', X.mean())
print('The maximum value in X is:', X.max())
print('The minimum value in X is:', X.min())
print('X has',(X < 0).sum(), 'negative numbers')
print('X has', (X > 0).sum(), 'positive numbers')
```
